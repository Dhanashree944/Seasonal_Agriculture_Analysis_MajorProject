**10**. **Outlier** **Analysis**

Potential outliers are examined using the IQR rule and boxplots. An extreme value is not removed automatically because it may represent a genuine agricultural situation.


```python
# IQR-based outlier summary for numerical columns

outlier_summary = []

for col in numeric_columns:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    count = ((df[col] < lower) | (df[col] > upper)).sum()

    outlier_summary.append({
        "Column": col,
        "Lower_Bound": lower,
        "Upper_Bound": upper,
        "Outlier_Count": count,
        "Outlier_Percentage": round(count / len(df) * 100, 2)
    })

outlier_summary = pd.DataFrame(outlier_summary).sort_values(
    "Outlier_Count", ascending=False
)

outlier_summary

```

```python
plt.figure(figsize=(8,5))
plt.boxplot(df["Yield_Tonnes_Ha"], vert=True)
plt.title("Yield Distribution and Potential Outliers")
plt.ylabel("Yield (tonnes/ha)")
plt.tight_layout()
plt.show()
```

**13**. **Seasonal** **Comparison**

The main seasonal metrics are summarized below so that Kharif, Rabi, and Zaid can be compared on the same basis.


```python
season_metrics =df.groupby("Season").agg(
    Records=("Farm_ID", "count"),
    Average_Yield=("Yield_Tonnes_Ha", "mean"),
    Total_Production=("Production_Tonnes", "sum"),
    Average_Profit=("Profit_INR", "mean"),
    Total_Profit=("Profit_INR", "sum"),
    Average_Water_Used=("Water_Used_m3", "mean")
).reindex(["Kharif", "Rabi", "Zaid"]).round(2)

season_metrics
```

```python
from matplotlib import colors
# 13.1 Average yield by season

season_colors = {
    "Kharif": "#2E7D32",   # Green
    "Rabi": "#1976D2",     # Blue
    "Zaid": "#F57C00"      # Orange
}

colors = [season_colors[season] for season in season_metrics.index]

plt.figure(figsize=(8,5))
plt.bar(
    season_metrics.index,
    season_metrics["Average_Yield"],
    color=colors
)

plt.title("Average Yield Across Seasons")
plt.xlabel("Season")
plt.ylabel("Average Yield (tonnes/ha)")
plt.tight_layout()
plt.show()
```

```python
# 13.2 Average profit by season

season_colors = {
    "Kharif": "#2E7D32",   # Green
    "Rabi": "#1976D2",     # Blue
    "Zaid": "#F57C00"      # Orange
}

colors = [season_colors[season] for season in season_metrics.index]

plt.figure(figsize=(8,5))
plt.bar(
    season_metrics.index,
    season_metrics["Average_Profit"],
    color=colors
)

plt.title("Average Profit Across Seasons")
plt.xlabel("Season")
plt.ylabel("Average Profit (INR)")
plt.tight_layout()
plt.show()
```

```python
# 13.3 Average water use by season

season_colors = {
    "Kharif": "#2E7D32",   # Green
    "Rabi": "#1976D2",     # Blue
    "Zaid": "#F57C00"      # Orange
}

colors = [season_colors[season] for season in season_metrics.index]

plt.figure(figsize=(8,5))
plt.bar(
    season_metrics.index,
    season_metrics["Average_Water_Used"],
    color=colors
)

plt.title("Average Water Usage Across Seasons")
plt.xlabel("Season")
plt.ylabel("Average Water Used (m³)")
plt.tight_layout()
plt.show()
```

**14. Additional Analysis**

```python
# 14.1 Regional performance
regional = df.groupby("State").agg(
    Farms=("Farm_ID","count"),
    Avg_Yield=("Yield_Tonnes_Ha","mean"),
    Avg_Profit=("Profit_INR","mean")
).sort_values("Avg_Profit", ascending=False).round(2)

regional
```

```python
plt.figure(figsize=(10,5))
top_states = regional.head(8).reset_index()
plt.bar(top_states["State"], top_states["Avg_Profit"])
plt.title("Average Profit by State (Top 8 by Profit)")
plt.xlabel("State")
plt.ylabel("Average Profit (INR)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# 14.2 Crop economics
crop_perf =df.groupby("Crop").agg(
    Avg_Yield=("Yield_Tonnes_Ha","mean"),
    Avg_Profit=("Profit_INR","mean"),
    Avg_Water=("Water_Used_m3","mean")
).sort_values("Avg_Profit", ascending=False).round(2)

crop_perf
```

```python
# 14.3 Resource efficiency by irrigation method
irrigation_perf = df.groupby("Irrigation_Method").agg(
    Avg_Yield=("Yield_Tonnes_Ha","mean"),
    Avg_Water=("Water_Used_m3","mean"),
    Avg_Water_Efficiency=("Water_Efficiency_t_per_1000m3","mean")
).sort_values("Avg_Water_Efficiency", ascending=False).round(2)

irrigation_perf
```

```python
plt.figure(figsize=(9,5))
sns.barplot(
    data=irrigation_perf.reset_index(),
    x="Irrigation_Method",
    y="Avg_Water_Efficiency",
    errorbar=None,
    color="steelblue"
)
plt.title("Average Water Efficiency by Irrigation Method")
plt.xlabel("Irrigation Method")
plt.ylabel("Water Efficiency (t per 1000 m³)")
plt.tight_layout()
plt.show()
```
